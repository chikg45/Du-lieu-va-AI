## Setup

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
from scipy import stats

sns.set_style('whitegrid')

csv_path = 'https://raw.githubusercontent.com/databricks/Spark-The-Definitive-Guide/master/data/retail-data/all/online-retail-dataset.csv'

df = pd.read_csv(csv_path, encoding='ISO-8859-1', on_bad_lines='skip')
goc = pd.read_csv(csv_path, encoding='ISO-8859-1', on_bad_lines='skip')

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print('Loaded from:', csv_path)
df.head()
```

## A.4. Create a new column
Làm sạch dữ liệu (loại Quantity<=0, UnitPrice<=0), tạo cột `Sales` = Quantity * UnitPrice.

```python
# TODO
df = df[(df['Quantity'] > 0) & (df["UnitPrice"] > 0)]
Sales = df["Quantity"] * df['UnitPrice']
df['Sales'] = Sales
```

# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO
print("Mean Mode Median")
for col in ['Quantity','UnitPrice','Sales']:
    print(col)
    print(f'Mean: {df[col].mean()} Mode: {df[col].mode().iloc[0]} Median: {df[col].median()}')
```

## Group 2 — Dispersion

```python
# TODO

for col in ['Quantity','UnitPrice','Sales']:
    print(col)
    q1,q2,q3 = df[col].quantile([0.25,0.5,0.75])
    print(f'var: {df[col].var()} , std: {df[col].std()}, diff: {df[col].max() -df[col].min()}, \nQuantile:  Q1: {q1}, Q2: {q2}, Q3: {q3} ')
```

## Group 3 — Location and Shape

```python
# TODO
for col in ['Quantity','UnitPrice','Sales']:   
    print(col)
    print(f"Skew: {df[col].skew():.4}, Kurt: {df[col].kurt():.4}")
```

## Câu hỏi 1: Quốc gia nào đóng góp doanh thu cao nhất, chiếm bao nhiêu % tổng doanh thu?

```python
# TODO
top = df.groupby('Country')['Sales'].sum().sort_values(ascending=False)
tong_doanh_thu = df['Sales'].sum()
qg_cao = top.idxmax()
gtri_cao = top.max()

print(f"Quoc gia co tong doanh thu cao nhat: {qg_cao}, voi ty le {(gtri_cao/tong_doanh_thu *100):.4}%")
```

## Câu hỏi 2: Sản phẩm nào bán chạy nhất theo doanh thu?

```python

df.groupby('Description')['Sales'].sum().sort_values(ascending=False)
```

## Câu hỏi 3: Doanh số có tính mùa vụ theo tháng không?

```python
# TODO
df['Month'] = df['InvoiceDate'].dt.month
monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales.sort_values(ascending=False))
print("Co tinh theo mua vu theo thang")
```
