## Phần 2: Lọc và sắp xếp dữ liệu


**Câu 6.** Lọc ra các đơn hàng có `SALES` lớn hơn 5000. Có bao nhiêu đơn hàng thỏa điều kiện này?

**Câu 7.** Lọc ra các đơn hàng thuộc quốc gia `USA` và có `DEALSIZE` là `Large`. Sắp xếp kết quả theo `SALES` giảm dần và hiển thị 10 dòng đầu tiên.

```python
sales_giam_dan = df[(df['COUNTRY'] == 'USA') & (df['DEALSIZE'] == "Large")].sort_values(by='SALES',ascending=False)
display(sales_giam_dan)
```

**Câu 8.** Lọc ra các đơn hàng có `Profit` âm (lỗ). Các đơn hàng này thường rơi vào `STATUS` nào? `DEALSIZE` nào? (Gợi ý: dùng `value_counts()`)

```python
display(df[df['Profit'] < 0]['DEALSIZE'].value_counts())
display(df[df['Profit'] < 0]['STATUS'].value_counts())
```

**Câu 9.** Lọc ra các đơn hàng có mức giảm giá (`Discount`) từ 0.3 trở lên. Doanh số (`SALES`) trung bình của nhóm này là bao nhiêu, so với doanh số trung bình toàn bộ dữ liệu?

```python
tbmau = df[df['Discount']>=0.3]['SALES'].mean()
tbtoandulieu= df['SALES'].mean()
print("Doanh so trung binh mau")
print(tbmau.round(4))
print('Doanh so trung binh toan du lieu')
print(tbtoandulieu.round(4))
print("Su chenh lech trung nhom nhom >0,3 so voi toan du lieu: ")
print("Chenh lech: ",(tbmau-tbtoandulieu).round(4))

```

## Phần 3: Tạo cột mới và biến đổi dữ liệu


**Câu 10.** Từ cột `ORDERDATE`, tạo thêm 3 cột mới: `ORDER_YEAR`, `ORDER_MONTH`, `ORDER_QUARTER` chứa năm, tháng, quý của ngày đặt hàng.

```python
df['ORDER_YEAR'] = df['ORDERDATE'].dt.year
# dt la datetime,l dung de lay thanh phan thoi gian, hoat dong khi cai cot muon lay phai la kieu datetime
df['ORDER_MONTH'] = df['ORDERDATE'].dt.month
df['ORDER_QUARTER'] = df['ORDERDATE'].dt.quarter
display(df)
```

**Câu 11.** Tạo cột mới `PROFIT_MARGIN` = `Profit` / `SALES` (tỷ suất lợi nhuận). Đơn hàng nào có tỷ suất lợi nhuận cao nhất? Thấp nhất?

```python
df['PROFIT_MARGIN'] = df['Profit']/df['SALES']
df

```

**Câu 12.** Tạo cột `NET_SALES` = `SALES` * (1 - `Discount`) thể hiện doanh số sau khi trừ giảm giá. So sánh tổng `SALES` và tổng `NET_SALES` của toàn bộ dữ liệu.

```python
df['NET_SALES'] = df['SALES']*(1-df['Discount'])
df
```

## Phần 4: Nhóm dữ liệu (Groupby) và tổng hợp


**Câu 13.** Tính tổng `SALES` và tổng `Profit` theo từng `COUNTRY`. Sắp xếp giảm dần theo tổng `SALES` và cho biết Top 5 quốc gia có doanh số cao nhất.

```python
print('Top 5 quoc gia co doanh so cao nhat')
df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False).head()
```

**Câu 14.** Tính `SALES` trung bình và `Profit` trung bình theo từng nhóm `DEALSIZE`. Nhận xét: nhóm `Large` có thực sự mang lại lợi nhuận trung bình cao nhất không?

```python
print('sale trung binh')
print(df.groupby('DEALSIZE')['SALES'].mean())
print('profit trung binh')
print(df.groupby('DEALSIZE')['Profit'].mean()
)

print("Nhan xet")
print('Nhom large khong mang lai gia tri trung binh lon nhat tai vi gia tri trung binh no chi la 24.04, Nhom Medium moi mang lai loi nhuan trung binh cao nhat 28,5')
```

**Câu 15.** Tính số lượng đơn hàng (đếm số `ORDERNUMBER` duy nhất) theo từng `STATUS`. Tỷ lệ đơn hàng bị `Cancelled` hoặc `Disputed` chiếm bao nhiêu % tổng số đơn hàng?

```python
df.groupby('STATUS')['ORDERNUMBER'].count()
```

**Câu 16.** Sử dụng `pivot_table` để tạo bảng tổng hợp tổng `SALES` theo `COUNTRY` (hàng) và `DEALSIZE` (cột).

```python
sales_pivot = df.pivot_table(index='COUNTRY',columns='DEALSIZE',values='SALES',aggfunc='sum')
print(sales_pivot)
```

**Câu 17.** Theo `ORDER_YEAR` đã tạo ở Câu 10, tính tổng doanh số (`SALES`) theo từng năm. Doanh số năm nào cao nhất? Lưu ý gì về dữ liệu năm 2005 (dữ liệu có đủ 12 tháng không)?

```python
print(df.groupby('ORDER_YEAR')['SALES'].sum())
print("Nam co doanh so cao nhat la nam : 2004")
print("Nam 2005 la nam co doanh thu thap nhat")
print("Kiem tra xem du lieu co du 12 thang khong( y la co thang nao khong co khach mua khong ?)")
print(df.groupby('ORDER_YEAR')['ORDER_MONTH'].unique().apply(sorted))
print("Nam 2005 chi co 5 thang, vi vay du lieu khong du 12 thang ")

```
