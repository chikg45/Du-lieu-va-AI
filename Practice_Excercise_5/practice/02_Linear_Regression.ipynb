# Bài tập Linear Regression

File nằm ở data/Salary_Data.csv.

Mỗi dòng là một người. YearsExperience là số năm kinh nghiệm, đây là feature. Salary là lương, đây là label. Label là số liên tục

Viết code theo từng ô bên dưới. Chạy từng ô một.

## Step 1. Load the data

Hint: import pandas, then use read_csv on data/Salary_Data.csv. Print the first 5 rows.

You should see two columns: YearsExperience and Salary.


```python
import pandas as pd

data = pd.read_csv("../data/Salary_Data.csv")
print(data.head())
```

## Step 2. Split feature and label

Hint: the feature is one column, YearsExperience. Keep it as a 2D table, like data[["YearsExperience"]]. The label is the Salary column.


```python
features = data[["YearsExperience"]]
label = data["Salary"]
```

## Step 3. Train the model

Hint: import LinearRegression from sklearn.linear_model. Create the model, then fit(features, label).

Print weight and bias:

- weight is model.coef_[0]. It is the slope: how much salary rises for each extra year of experience
- bias is model.intercept_. It is the predicted salary when years of experience are 0

Print this formula:

Salary = bias + weight * YearsExperience


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(features, label)

weight = model.coef_[0]
bias = model.intercept_

print("weight =", weight)
print("bias =", bias)
print(f"Salary = {bias:.2f} + {weight:.2f} * YearsExperience")
```

## Step 4. Predict

Hint: build a small table with YearsExperience equal to 5, then call model.predict. predict needs a 2D table, not a single number.

A reasonable result is around 73000.


```python
years = pd.DataFrame({"YearsExperience": [5]})
salary = model.predict(years)[0]
print("Predicted salary for 5 years of experience =", round(salary, 2))
```
