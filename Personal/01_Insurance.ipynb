## Group 1 — Central Tendency

```python
# TODO
print("Mean Mode Median")
for col in ['bmi','charges']:
    print(col)
    print(f'Mean: {df[col].mean()} Mode: {df[col].mode().iloc[0]} Median: {df[col].median()}')
```

```python
# TODO

for col in ['bmi','charges']:
    print(col)
    q1,q2,q3 = df[col].quantile([0.25,0.5,0.75])
    print(f'var: {df[col].var()} , std: {df[col].std()}, diff: {df[col].max() -df[col].min()}, \nQuantile:  Q1: {q1}, Q2: {q2}, Q3: {q3} ')
```

## Câu hỏi 1: Người hút thuốc trả chi phí cao hơn bao nhiêu lần so với người không hút, và có đồng đều giữa các vùng không?

```python
# TODO
avg = df.groupby('smoker')['charges'].mean()
chiphi = avg['yes'] / avg['no']

print(f'Người hút thuốc có chi phí cao hơn {chiphi:.2f} lần')
region_smoker = df.groupby(['region', 'smoker'])['charges'].mean()

print(region_smoker)
region_ratio = (
    df.groupby(['region', 'smoker'])['charges']
      .mean()
      .unstack()
)

region_ratio['ratio'] = region_ratio['yes'] / region_ratio['no']

print(region_ratio)
print("Nhin chung khong co su dong deu giua cac vung")
```

## Câu hỏi 2: BMI có tương quan với chi phí mạnh hơn ở nhóm hút thuốc hay không hút thuốc?

```python
# TODO
corr = df.groupby('smoker').apply(
    lambda x: x['bmi'].corr(x['charges'])
)

print(corr)
```

## Câu hỏi 3: Vùng nào có chi phí bảo hiểm trung bình cao nhất?

```python
# TODO
avg_charges_region = df.groupby('region')['charges'].mean()

print(avg_charges_region)
```

## Câu hỏi 4: Số lượng con cái có làm tăng chi phí bảo hiểm không?

```python
# TODO
avg_charges = df.groupby('children')['charges'].mean().sort_values(ascending=False)

print(avg_charges)
print("So luong con cai nhin chung co lam tang chi phi bao hiem")
```
