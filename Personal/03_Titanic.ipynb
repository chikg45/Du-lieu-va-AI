# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO
df.describe()
print("Mean Median Mode")
for col in ['fare']:
    mean_, median_, mode_ = df[col].mean(), df[col].median(), df[col].mode()[0]
    print(f"{col}: Mean={mean_:.2f}, Median={median_:.2f}, Mode={mode_:.2f}")

```

## Group 2 — Dispersion

```python
# TODO
for col in ['fare']:
    s = df[col]
    range_ = s.max()-s.min()
    std_ = s.std()
    q1,q3 = s.quantile([0.25,0.75]); iqr = q3-q1
    cv = std_/s.mean()
    print(f"--- {col} ---")
    print(f"Range={range_:.2f}, Std={std_:.2f}, IQR={iqr:.2f}, CV={cv:.2f}")
```

## Group 3 — Location and Shape

```python
# TODO
for col in ["fare"]:
    s = df[col]
    skew_,kurt_ = stats.skew(s),stats.kurtosis(s)
    print("Skew va Kurt cua fare")
    print(f"Skew: {skew_.round(4)}, Kurt: {kurt_.round(4)}")
print("Skew > 0 ~~ 4,8 , lech nhieu ve ben phai")
print("có kurtosis rất cao, nghĩa là phân phối có đuôi rất nặng và xuất hiện nhiều giá trị cực đoan so với phân phối chuẩn.")
```
